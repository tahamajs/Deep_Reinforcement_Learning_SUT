# CA19 Demo

This notebook demonstrates a short debug run of the Actor-Critic Ensemble from `src/`. It runs a small training loop (no GPU required) and saves a reward plot to `pictures/fig_01_reward.png`.


```python
# Setup: imports and path
from pathlib import Path
import sys

# ensure demo can import local src when running from the notebooks folder
sys.path.append(str(Path("..").resolve() / "src"))

from config import CAConfig
from train import run_training

cfg = CAConfig()
# use small debug settings
cfg.seed = 42
cfg.total_steps = 200
cfg.batch_size = 8
cfg.lr = 1e-3
cfg.env_name = "CartPole-v1"
cfg.device = "cpu"
```

```python
# Run a short training session (this cell may take a minute)
out = run_training(cfg)
print("Episodes:", len(out["rewards"]))
print("Steps:", out["steps"])
print("Checkpoint:", out["checkpoint"])

# The reward plot is saved to ../pictures/fig_01_reward.png
print("Saved plot at: ../pictures/fig_01_reward.png")
```
