# AU-DMG demo

This notebook loads training logs and evaluation artifacts and reproduces key figures.


```python
import csv
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

log = Path("../outputs/ca9").glob("train_*.csv")
log = sorted(list(log))
assert len(log) > 0, "no logs found in ../outputs/ca9/"
path = log[-1]
rows = []
with open(path) as f:
    reader = csv.DictReader(f)
    for r in reader:
        rows.append(r)
steps = [int(r["step"]) for r in rows]
v = [float(r["v_loss"]) for r in rows]
c = [float(r["critic_loss"]) for r in rows]
p = [float(r["policy_loss"]) for r in rows]
lam = [float(r["lam_mean"]) for r in rows]
std = [float(r["std_mild_mean"]) for r in rows]

plt.figure(figsize=(8, 4))
plt.plot(steps, v, label="value")
plt.plot(steps, c, label="critic")
plt.plot(steps, p, label="policy")
plt.legend()
plt.title("Losses")
plt.xlabel("step")
plt.show()
```

```python
plt.figure(figsize=(8, 4))
plt.plot(steps, lam, label="Lambda")
plt.plot(steps, std, label="Std")
plt.legend()
plt.title("Lambda and Std")
plt.xlabel("step")
plt.show()
```

```python
import numpy as np

ep0 = Path("../outputs/ca9/eval/ep_0/rews.npy")
if ep0.exists():
    rews = np.load(ep0, allow_pickle=True)
    plt.figure(figsize=(8, 4))
    plt.plot(np.arange(len(rews)), rews)
    plt.title("Episode 0 rewards")
    plt.xlabel("step")
    plt.show()
else:
    print("No eval episode found at", ep0)
```
