# MaxSink visualization demo

This notebook demonstrates plotting utilities for reward histograms, Sinkhorn loss curves, and particle PCA scatter. It imports `visualize.py` from the CA8 package.


```python
import sys
from pathlib import Path

nb_root = Path("..")
src_dir = nb_root.parent / "src"
if str(src_dir) not in sys.path:
    sys.path.insert(0, str(src_dir))

from visualize import plot_reward_histograms, plot_sinkhorn_loss, plot_particle_pca
import numpy as np

# Synthetic demo data
rng = np.random.RandomState(1)
raw = rng.choice([0.0, 1.0], size=1000, p=[0.95, 0.05])
transformed = np.clip(
    raw + rng.normal(scale=0.1, size=raw.shape) + (raw > 0) * 0.4, 0.0, 1.4
)
plot_reward_histograms(
    raw, transformed, save_path="../pictures/demo_reward_hist_nb.png"
)
losses = np.abs(np.sin(np.linspace(0, 6.0, 500))) + np.linspace(0.2, 0.01, 500)
plot_sinkhorn_loss(losses, save_path="../pictures/demo_sinkhorn_loss_nb.png")
pred = rng.normal(size=(200, 32, 1))
targ = pred + rng.normal(scale=0.3, size=pred.shape)
plot_particle_pca(pred, targ, save_path="../pictures/demo_particles_pca_nb.png")
print("Saved demo images to ../pictures/")
```
