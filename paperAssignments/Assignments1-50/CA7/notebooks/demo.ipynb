## 3. Plotting ideas

Below are example matplotlib calls to plot returns and trace norms from logs (assumes a `logs.csv` exists).


```python
import matplotlib.pyplot as plt
import pandas as pd

def plot_returns(log_csv='../logs.csv'):
    df = pd.read_csv(log_csv)
    plt.figure(figsize=(6,4))
    plt.plot(df['step'], df['return_mean'], label='return')
    plt.fill_between(df['step'], df['return_mean']-df['return_std'], df['return_mean']+df['return_std'], alpha=0.2)
    plt.xlabel('steps')
    plt.ylabel('return')
    plt.legend()
    plt.savefig('../pictures/fig_returns.png', dpi=300)
```
