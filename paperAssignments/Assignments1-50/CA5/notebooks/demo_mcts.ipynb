This notebook demonstrates the PUCT MCTS implementation against a tiny 1D toy model. The code cells are written so they can be run directly in a Jupyter environment. The notebook intentionally contains no outputs.


```python
import sys
from pathlib import Path

repo_root = Path(__file__).parents[2]
sys.path.append(str(repo_root))

import matplotlib.pyplot as plt
from src.mcts.puct import PUCT
from src.crosshq.model import GaussianPolicy
from scripts.demo_mcts import ToyModel, run_demo

# Run the demo search
root, best_idx = run_demo(num_simulations=100, verbose=True)

# Plot visit distribution (if available)
visits = {a: child.visits for a, child in root.children.items()}
labels = [str(a) for a in sorted(visits.keys())]
counts = [visits[a] for a in sorted(visits.keys())]
plt.figure(figsize=(4, 3))
plt.bar(labels, counts)
plt.xlabel("Action index")
plt.ylabel("Visit counts")
plt.title("PUCT Visit Counts (root)")
plt.show()
```
