# TWM-SSD Demo Notebook
+This notebook shows how to import and instantiate the TWM-SSD model (no execution required).

```python
import torch
from paperAssignments.Assignments1_50.CA11.src.config import get_default_config
from paperAssignments.Assignments1_50.CA11.src.model import TWMSSDModel
from paperAssignments.Assignments1_50.CA11.src.data import RandomTrajectoryDataset

cfg = get_default_config()
model = TWMSSDModel(d_model=cfg.d_model, n_heads=cfg.n_heads, n_layers=cfg.n_layers)
print('Model created:', type(model))

# Example forward (shapes only)
batch = 2
obs = torch.randn(batch, cfg.seq_len, cfg.d_model)
acts = torch.randn(batch, cfg.seq_len, cfg.d_model)
pred_obs, pred_reward = model(obs, acts)
print('pred_obs.shape =', pred_obs.shape)
print('pred_reward.shape =', pred_reward.shape)
```
