## Simulation Tools


Both the `angle_normalize` function and the `PendulumDynamics` class are fundamental components for accurately simulating, analyzing, and controlling the pendulum system.
They ensure consistency in angle representation and provide a realistic model of the pendulum's behavior, enabling effective control strategies.

We use the `angle_normalize` function for:

- **Consistency**: When dealing with angles, it's important to keep them within a standard range to ensure consistency in calculations.
- **Handling Wrapping**: Angles can wrap around when they exceed $2\pi$ or drop below $-2\pi$. Normalizing angles helps avoid confusion and errors that can arise from angle wrapping.

And we use the `PendulumDynamics` class for:

- **Modeling Physical Behavior**: The `PendulumDynamics` class models the physical behavior of the pendulum.
- **Simulation and Control**: This class allows us to simulate the pendulum's response to different actions, which is crucial for designing and testing control algorithms.
- **Optimization**: Understanding the dynamics of the pendulum helps in optimizing the control inputs. The class encapsulates the physics involved, enabling us to apply control techniques like Model Predictive Control (MPC) to achieve the desired behavior.


```python
class PendulumDynamics(nn.Module):
    def forward(self, state, action):
        th = state[:, 0].view(-1, 1)
        thdot = state[:, 1].view(-1, 1)

        g = 10  # default value of the environment (not 9.81)
        m = 1
        l = 1
        dt = 0.05

        u = action
        u = torch.clamp(u, -2, 2)

        newthdot = (
            thdot
            + (-3 * g / (2 * l) * torch.sin(th + np.pi) + 3.0 / (m * l**2) * u) * dt
        )
        newth = th + newthdot * dt
        newthdot = torch.clamp(newthdot, -8, 8)

        state = torch.cat((angle_normalize(newth), newthdot), dim=1)
        return state


def angle_normalize(x):
    return ((x + math.pi) % (2 * math.pi)) - math.pi
```

## The Cost Function


The `define_swingup_goal` function creates a cost function that the MPC algorithm uses to determine the optimal control actions to achieve the desired pendulum swing-up task.
It considers both the desired state (upright and stationary) and penalizes large control inputs to ensure smooth control actions.


```python
def define_swingup_goal():
    goal_weights = torch.tensor((1.0, 0.1))  # Weights for theta and theta_dot
    goal_state = torch.tensor((0.0, 0.0))  # Desired state (theta=0, theta_dot=0)
    ctrl_penalty = 0.001
    q = torch.cat((goal_weights, ctrl_penalty * torch.ones(1)))  # Combined weights
    px = -torch.sqrt(goal_weights) * goal_state
    p = torch.cat((px, torch.zeros(1)))
    Q = torch.diag(q).repeat(TIMESTEPS, N_BATCH, 1, 1)  # Cost matrix
    p = p.repeat(TIMESTEPS, N_BATCH, 1)
    return mpc.QuadCost(Q, p)  # Quadratic cost
```

## Running MPC


To run the MPC, in each iteration:

1.  First you obtain the current state of the environment and convert it to a tensor.
2.  Then you recreate the MPC controller using the updated `u_init` and calculate the optimal control actions based on the current state, dynamics, and cost function.
3.  Next you take the first planned action and update `u_init` with the remaining actions.
4.  Finally you take a step in the environment and store the rewards and actions.

Remember that `u_init` serves as the initial guess for the control inputs.


```python
# Select the hyperparameters
RUN_ITER = 200  # Number of iterations to run MPC
TIMESTEPS = 10  # MPC horizon length
N_BATCH = 1  # Batch size for MPC
LQR_ITER = 5  # Number of LQR iterations for convergence


# Setup the environmnet
env = gym.make("Pendulum-v1", render_mode="rgb_array")
video_directory = "mpc_videoss"
env = RecordVideo(env, video_directory)
env.reset()  # Reset the underlying environment
env.unwrapped.state = [np.pi, 1]  # Environment must start in downward position


# Define the cost function and initialize u
cost = define_swingup_goal()
u_init = None


# Run MPC
rewards, actions = [], []
for _ in (pbar := trange(RUN_ITER)):
    state = env.unwrapped.state.copy()
    state = torch.tensor(state).view(1, -1)
    # recreate controller using updated u_init (kind of wasteful right?)
    ctrl = mpc.MPC(
        2,
        1,
        TIMESTEPS,
        u_lower=-2.0,
        u_upper=+2.0,
        lqr_iter=LQR_ITER,
        exit_unconverged=False,
        eps=1e-2,
        n_batch=N_BATCH,
        backprop=False,
        verbose=0,
        u_init=u_init,
        grad_method=mpc.GradMethods.AUTO_DIFF,
    )

    # compute action based on current state, dynamics, and cost
    nominal_states, nominal_actions, nominal_objs = ctrl(
        state, cost, PendulumDynamics()
    )

    # Take first planned action
    action = nominal_actions[0, 0, 0].item()

    # Update u_init
    u_init = torch.cat((nominal_actions[1:], torch.zeros(1, N_BATCH, 1)), dim=0)

    # Take a step in the environment
    state, reward, terminated, truncated, info = env.step([action])

    # Store the latest action and reward
    actions.append(action)
    rewards.append(reward)

    pbar.set_description(f"Action = {actions[-1]:.2f} | Reward = {rewards[-1]:.2f}")
    env.render()

env.close()

# Plot the results
plot_results(rewards, actions)
```

`Your Answers:`

**1. How does the number of LQR iterations affect the MPC?**

The number of LQR iterations determines how well the MPC controller converges to the optimal solution. More LQR iterations generally lead to better convergence and more accurate control actions, but at the cost of increased computational time. With too few iterations, the controller might not converge properly and could produce suboptimal actions. However, there's a diminishing returns effect - beyond a certain point, additional iterations provide minimal improvement.

**2. What if we didn't have access to the model dynamics? Could we still use MPC?**

No, traditional MPC requires an accurate model of the system dynamics to predict future states and optimize control actions. Without the model dynamics, we cannot:

- Predict how the system will evolve under different control inputs
- Solve the optimization problem that MPC relies on
- Plan ahead effectively

However, we could potentially use data-driven approaches like:

- Learning the dynamics model from data (model learning)
- Using neural networks to approximate the dynamics
- Model-free reinforcement learning methods instead

**3. Do `TIMESTEPS` or `N_BATCH` matter here? Explain.**

Yes, both parameters significantly affect MPC performance:

- **TIMESTEPS**: This is the prediction horizon length. A longer horizon allows the controller to plan further ahead, which is crucial for tasks like swing-up where the pendulum needs to build momentum over multiple steps. However, longer horizons increase computational cost and may lead to less accurate long-term predictions.

- **N_BATCH**: This determines how many parallel optimization problems are solved. In this single-agent case, N_BATCH=1 is appropriate. Larger batch sizes would be useful for multi-agent scenarios or when solving multiple related control problems simultaneously, but they increase computational requirements.

**4. Why do you think we chose to set the initial state of the environment to the downward position?**

We set the initial state to the downward position (θ=π, θ̇=1) because:

- This represents the most challenging starting condition for the swing-up task
- It tests the controller's ability to perform the complete swing-up maneuver from the worst-case scenario
- Starting from rest (θ=π, θ̇=0) would be easier but less representative of real-world scenarios
- It demonstrates the controller's capability to handle the full range of the pendulum's motion

**5. As time progresses (later iterations) what happens to the actions and rewards? Why?**

As the MPC controller progresses:

- **Actions**: Initially, actions are large and aggressive to build momentum for swing-up. As the pendulum approaches the upright position, actions become smaller and more precise for stabilization. Eventually, actions oscillate around small values to maintain balance.

- **Rewards**: Rewards start very negative (around -16) due to the pendulum being far from the goal state. As the controller successfully swings up the pendulum, rewards improve significantly. Once the pendulum is balanced upright, rewards stabilize around higher values (closer to 0), with small fluctuations due to the continuous control needed to maintain balance.

This behavior occurs because the MPC controller is optimizing for the swing-up and balance task, transitioning from large corrective actions to fine-tuning control as the system approaches the desired state.
