## 1. Introduction

Multi-modal agents must fuse heterogeneous signals while avoiding spurious correlations. Causal structure and counterfactual reasoning improve robustness and transfer. This notebook frames CA8 with the essential causal + fusion tools.

## 2. Methodology

- **Causal modeling:** SCMs with do-calculus; back-door/front-door adjustment for policy evaluation; intervention-aware dynamics.
- **Counterfactual/importance corrections:** Doubly robust or marginalized IS for offline data; causal world models for planning.
- **Invariant objectives:** IRM/IRMv1 or risk extrapolation across environments to prefer causal features.
- **Multi-modal fusion:** Modality-specific encoders (CNN/Transformer/RNN) with cross-attention or gating; late vs intermediate fusion; modality dropout for robustness.
- **Representation learning:** Contrastive/InfoNCE across modalities; disentangled latents; temporal alignment for sequences.
- **Safety & fairness:** Monitor distribution shift; enforce constraints when intervening; document assumptions on causal graph.

## 3. Experiments (design)

- Compare fusion strategies (late vs cross-attention) under missing/noisy modalities.
- Evaluate causal vs non-causal objectives on shifted test distributions; offline vs online data.
- Ablate adjustment sets, IRM penalties, and counterfactual estimators; measure return, calibration, and OOD gap.
- Datasets/environments: multi-modal control or gridworld with spurious cues; fixed seeds; small rollout budgets for reproducibility.

## 4. Conclusion

Causal structure plus resilient fusion yields policies that transfer and resist spurious correlations. Counterfactual evaluation and invariant objectives guard against distribution shift; modality-aware encoders handle missing or noisy inputs.
